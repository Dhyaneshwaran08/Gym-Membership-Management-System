```python
# ============================================================
# GYM MEMBERSHIP MANAGEMENT SYSTEM
# Developed using Python
# ============================================================

print("=" * 55)
print("          GYM MEMBERSHIP MANAGEMENT SYSTEM")
print("=" * 55)

# ------------------------------------------------------------
# 1. MEMBER DETAILS
# ------------------------------------------------------------

Member_Name = input("Enter Member Name: ")
Member_ID = int(input("Enter Member ID: "))
Mobile_Number = input("Enter Mobile Number: ")
Membership_Plan = input("Enter Membership Plan (Basic/Gold/Premium): ")
Number_of_Months = int(input("Enter Number of Months: "))
Number_of_services_used = int(input("Enter Number of Services Used: "))
Membership_Status = input("Enter Membership Status (Active/Inactive): ")

print("\n--- MEMBER INFORMATION ---")
print("Member Name       :", Member_Name)
print("Member ID         :", Member_ID)
print("Mobile Number     :", Mobile_Number)
print("Membership Plan   :", Membership_Plan)
print("Number of Months  :", Number_of_Months)
print("Services Used     :", Number_of_services_used)
print("Membership Status :", Membership_Status)


# ------------------------------------------------------------
# 2. SERVICE COST MANAGEMENT USING LIST
# ------------------------------------------------------------

Service_Cost = []
Total = 0

print("\n--- ENTER SERVICE COSTS ---")

for i in range(Number_of_services_used):

    Cost = float(input(f"Enter Cost of Service {i + 1}: ₹"))

    # Prevent negative service cost
    while Cost < 0:
        print("Service cost cannot be negative.")
        Cost = float(input("Enter Service Cost Again: ₹"))

    Service_Cost.append(Cost)
    Total += Cost


print("\nService Costs:", Service_Cost)


# ------------------------------------------------------------
# 3. LIST OPERATIONS
# ------------------------------------------------------------

if Service_Cost:

    print("\n--- SERVICE COST SUMMARY ---")

    Service_Cost.sort()

    print("Sorted Service Costs :", Service_Cost)
    print("Number of Services   :", len(Service_Cost))
    print("Highest Service Cost : ₹", max(Service_Cost))
    print("Lowest Service Cost  : ₹", min(Service_Cost))
    print("Total Service Cost   : ₹", sum(Service_Cost))

else:

    print("\nNo services were added.")


# ------------------------------------------------------------
# 4. OPTIONAL ADDITIONAL SERVICES
# ------------------------------------------------------------

Additional_Services = int(
    input("\nEnter Number of Additional Services (0 if none): ")
)

for i in range(Additional_Services):

    print(f"\nAdditional Service {i + 1}")

    Cost_1 = float(
        input("Enter Service Cost (0 = Skip, 9999 = Stop): ₹")
    )

    while Cost_1 < 0:
        print("Negative cost is not allowed.")
        Cost_1 = float(input("Enter Cost Again: ₹"))

    if Cost_1 == 0:
        print("Service Skipped")
        continue

    if Cost_1 == 9999:
        print("Additional Service Entry Stopped")
        break

    Service_Cost.append(Cost_1)
    Total += Cost_1


print("\nUpdated Service Costs:", Service_Cost)


# ------------------------------------------------------------
# 5. TUPLE OPERATIONS
# ------------------------------------------------------------

Cost_Tuple = tuple(Service_Cost)

print("\n--- SERVICE COST HISTORY ---")
print("Cost Tuple:", Cost_Tuple)

if Cost_Tuple:

    print("First Service Cost :", Cost_Tuple[0])
    print("Last Service Cost  :", Cost_Tuple[-1])
    print("First 3 Costs      :", Cost_Tuple[:3])
    print("Last 2 Costs       :", Cost_Tuple[-2:])
    print("Reversed Costs     :", Cost_Tuple[::-1])

    print("\nIndividual Service Costs:")

    for i in range(len(Cost_Tuple)):
        print(f"Service {i + 1}: ₹{Cost_Tuple[i]}")


# ------------------------------------------------------------
# 6. MEMBER DETAILS USING DICTIONARY
# ------------------------------------------------------------

Member_Details = {
    "Member Name": Member_Name,
    "Member ID": Member_ID,
    "Mobile Number": Mobile_Number,
    "Membership Plan": Membership_Plan,
    "Membership Status": Membership_Status,
    "Number of Months": Number_of_Months
}

Member_Details["Total Service Cost"] = Total

if len(Service_Cost) > 0:

    Member_Details["Average Service Cost"] = (
        Total / len(Service_Cost)
    )

    Member_Details["Highest Service Cost"] = max(Service_Cost)

else:

    Member_Details["Average Service Cost"] = 0
    Member_Details["Highest Service Cost"] = 0


Gym_Name = input("\nEnter Gym Name: ")

Member_Details["Gym Name"] = Gym_Name


print("\n--- MEMBER DETAILS ---")

for Key, Value in Member_Details.items():
    print(f"{Key}: {Value}")


# ------------------------------------------------------------
# 7. DICTIONARY OPERATIONS
# ------------------------------------------------------------

print("\nDictionary Keys:")
print(Member_Details.keys())

print("\nDictionary Values:")
print(Member_Details.values())

print("\nDictionary Items:")
print(Member_Details.items())


# ------------------------------------------------------------
# 8. SERVICE BENEFITS - NESTED LOOP
# ------------------------------------------------------------

Number_of_Benefit_Services = int(
    input("\nEnter Number of Services to Add Benefits For: ")
)

for i in range(Number_of_Benefit_Services):

    print(f"\nService {i + 1}")

    Number_of_Benefits = int(
        input("Enter Number of Benefits: ")
    )

    for j in range(Number_of_Benefits):

        Benefit = input(
            f"Enter Benefit {j + 1}: "
        )

        print("Benefit Added:", Benefit)


# ------------------------------------------------------------
# 9. BILL CALCULATION
# ------------------------------------------------------------

Registration_Charge = 1000

Total_Cost = Total + Registration_Charge

print("\n--- BILL CALCULATION ---")

print("Service Cost        : ₹", Total)
print("Registration Charge : ₹", Registration_Charge)
print("Subtotal            : ₹", Total_Cost)


# Discount Calculation

Discount = 0

if Total_Cost > 3000:

    Discount = 1000
    print("Discount Applied    : ₹", Discount)


Discounted_Amount = Total_Cost - Discount


# GST after discount

GST = Discounted_Amount * 18 / 100

Final_Amount = Discounted_Amount + GST


print("Amount After Discount: ₹", round(Discounted_Amount, 2))
print("GST (18%)            : ₹", round(GST, 2))
print("Final Amount         : ₹", round(Final_Amount, 2))


# ------------------------------------------------------------
# 10. STRING OPERATIONS
# ------------------------------------------------------------

print("\n--- STRING OPERATIONS ---")

if Member_Name:

    Indexing = Member_Name[0]
    print("First Character:", Indexing)


Concatenation = Member_Name + " - " + Membership_Status
print("Concatenation:", Concatenation)

Replication = Member_Name * 2
print("Replication:", Replication)

Membership_Operator = "a" in Member_Name.lower()
print("Does Name Contain 'a'?:", Membership_Operator)


# ------------------------------------------------------------
# 11. PYTHON OPERATORS
# ------------------------------------------------------------

print("\n--- OPERATOR DEMONSTRATION ---")

Unit_Digit = Member_ID % 10
print("Last Digit of Member ID:", Unit_Digit)


Statement_1 = Number_of_Months
Statement_2 = Number_of_services_used


# Relational Operator

print(
    "Months >= Services:",
    Statement_1 >= Statement_2
)


# Logical Operator

print(
    "Eligible Based on Duration & Services:",
    Statement_1 >= 3 and Statement_2 >= 2
)


# Bitwise Operator - included to demonstrate Python bitwise operation

print(
    "Bitwise OR Result:",
    Statement_1 | Statement_2
)


# ------------------------------------------------------------
# 12. VARIABLE SWAPPING
# ------------------------------------------------------------

a = Number_of_Months
b = Number_of_services_used

a, b = b, a

print("\n--- VARIABLE SWAPPING ---")
print("After Swapping:", a, b)


# ------------------------------------------------------------
# 13. MEMBERSHIP CATEGORY
# ------------------------------------------------------------

print("\n--- MEMBERSHIP CATEGORY ---")

if Final_Amount >= 3500:

    Member_Category = "Premium Member"

elif Final_Amount >= 2500:

    Member_Category = "Gold Member"

else:

    Member_Category = "Basic Member"


print("Member Category:", Member_Category)


# ------------------------------------------------------------
# 14. MEMBERSHIP BENEFIT ELIGIBILITY
# ------------------------------------------------------------

print("\n--- BENEFIT ELIGIBILITY ---")

if Membership_Status.lower() == "active":

    if Final_Amount >= 3500:

        if Number_of_Months >= 3:

            print(
                "Eligible for Free Personal Training Session"
            )

        else:

            print(
                "Complete at least 3 months to receive "
                "the training benefit."
            )

    else:

        print(
            "Premium spending criteria not met."
        )

else:

    print(
        "Membership is inactive. Benefits are unavailable."
    )


# ------------------------------------------------------------
# 15. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("                 FINAL MEMBER SUMMARY")
print("=" * 55)

print("Gym Name          :", Gym_Name)
print("Member Name       :", Member_Name)
print("Member ID         :", Member_ID)
print("Membership Plan   :", Membership_Plan)
print("Membership Status :", Membership_Status)
print("Membership Period :", Number_of_Months, "Months")
print("Total Services    :", len(Service_Cost))
print("Service Cost      : ₹", round(Total, 2))
print("Discount          : ₹", round(Discount, 2))
print("GST               : ₹", round(GST, 2))
print("Final Amount      : ₹", round(Final_Amount, 2))
print("Member Category   :", Member_Category)

print("=" * 55)
print("          THANK YOU FOR USING THE SYSTEM")
print("=" * 55)
```
